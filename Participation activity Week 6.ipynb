write a lambda function that takes the Student Performance Data Set and returns another data frame.

```python
import pandas as pd
import io
from google.colab import drive
drive.mount('/content/drive')
```

```python
df = pd.read_csv('/content/drive/MyDrive/MSSP607/Student Performance Data.csv')
```

```python
df['avg_score'] = df.apply(lambda x: (x['math score'] + x['reading score'] + x['writing score'])/3, axis = 1)
```

```python
df['pass_fail'] = df.apply(lambda x: 'Pass' if x['avg_score'] >= 60 else 'Fail', axis = 1)
```

```python
gender_col = None
for col in df.columns:
  if col.lower() == 'gender':
    gender_col = col
    break
new_df = df[[gender_col, 'math score', 'reading score', 'writing score', 'avg_score', 'pass_fail']]
```
