d. Add a column to the dataframe indicating whether or not the earthquake happened in a country or US state that is on the Ring of Fire.
```python
Ring_of_Fire = ['Bolivia','Chile', 'Ecuador', 'Peru', 'Costa Rica', 'Guatemala', 'Mexico'
'Japan', 'Philippines', 'Indonesia', 'New Zealand', 'Antarctica','Japan', 'Philippines',
'Indonesia', 'New Zealand', 'Antarctica', 'Canada', 'Fiji', 'Alaska', 'Washington',
'California', 'Russia', 'Taiwan', 'Tonga', 'Kermadec Islands']
```

```python
df['Ring_of_Fire'] = df['place'].apply(lambda x:any(loc.lower() in str(x).lower() for loc in Ring_of_Fire))
```
